# 1. Pre-Processing

## 1.1 Import necessary Library

```python
import pandas as pd
import numpy as np
```

## 1.2 Load CSV

```python
literacy = pd.read_csv('../data/table17_literacy.csv')
attainment = pd.read_csv('../data/table18_attainment.csv')
attendance= pd.read_csv('../data/table20_attendance.csv')

print("Table 17 shape:", literacy.shape)
print("Table 18 shape:", attainment.shape)
print("Table 20 shape:", attendance.shape)
```

## 1.2 Inspect the Data

```python
# Print out all column names of all three csv
print(literacy.columns.tolist())
print(attainment.columns.tolist())
print(attendance.columns.tolist())
```

### 1.2.1 Identifying merging columns

literacy and attainment both have column 'sex' with value male=1 and female=2, but attendence have seperate  column for gender(eg. male, female, attnding_m, attnding_f).

Merging all df on common keys ['prov','dist','gapa','sex','agegrp'] will fail because attendance has no 'sex' column to merge on

Fix:
Reshape attendance from wide format to long format.
so instead of multiple columns for gender, we have only one column.

### 1.2.2 Reshaping Attendance from wide to long format

```python
# reshaping 
# combine all male into a new df
attendance_male = attendance[['prov', 'dist', 'gapa', 'agegrp', 'male', 'attnding_m', 'evrattnding_m', 'nvrattnding_m', 'notstd_m']]
print(attendance_male.head())
attendance_female = attendance[['prov', 'dist', 'gapa', 'agegrp', 'female', 'attnding_f', 'evrattnding_f', 'nvrattnding_f', 'notstd_f']]
print(attendance_female.head())
```

### 1.2.3 Adding a New column named sex

```python
# male and female became rowtotal_20 to keep the total value, a new column is added to preserve the gender data
attendance_male.columns = ['prov', 'dist', 'gapa', 'agegrp', 'rowtotal_20', 'attnding', 'evrattnding', 'nvrattnding', 'notstd_20']
attendance_male['sex'] = 1

attendance_female.columns = ['prov', 'dist', 'gapa', 'agegrp', 'rowtotal_20', 'attnding', 'evrattnding', 'nvrattnding', 'notstd_20']
attendance_female['sex'] = 2
```

```python
 # check
print(attendance_male.head())
print(attendance_female.head())
```

### 1.2.4 Stack male and female datafrmae as one


```python

attendance_gender=pd.concat([attendance_male,attendance_female],ignore_index=True)
print("Shape of attendance gender", attendance_gender.shape)
attendance_gender
```

## 1.3 Another complication

1. Merging literacy and attendance_gender on common keys
2. Calculating attendance rate(attending/rowtotal), in theory it shoulndnt exceed 1.
3. making a new df called invalid rows which consists rows whose rate is > 1 to identify problem

```python
# Test merge literacy and attendance_gender on common keys
# merging literacy and attendance_gender on common keys
merge_test = pd.merge(literacy, attendance_gender, on=['prov','dist','gapa','sex','agegrp'])
print("Merged shape:", merge_test.shape)

#Calculate attendance rate that should be between 0 and 1
merge_test['attending_rate'] = merge_test['attnding'] / merge_test['rowtotal']
print(merge_test['attending_rate'].describe())

# Isolate rows where the rate is impossible (above 1)
invalid_rows = merge_test[merge_test['attending_rate'] > 1]
print("Number of rows with rate > 1:", invalid_rows.shape[0])
invalid_rows[['prov','dist','gapa','sex','agegrp','agegrpname','rowtotal','attnding']].head(10)
```

```python
print("Literacy agegrp mapping:")
print(literacy[['agegrp','agegrpname']].drop_duplicates().sort_values('agegrp'))

print("\nAttendance (original) agegrp mapping:")
print(attendance[['agegrp','agegrpname']].drop_duplicates().sort_values('agegrp'))
print("\nAttainment agegrp mapping:")
print(attainment[['agegrp','agegrpname']].drop_duplicates().sort_values('agegrp'))
```


### 1.3.1 Problem Identified

#### 1.3.1.1 agegrp problem
Looking at the invalid rows, agegrp=7 is labeled "11 Year" in literacy. This looks inconsistent compared to waht i initially viewed in csv to check data. code 7 should logically mean "7 years old" if the codes directly matched ages.

To confirm, the full agegrp-to-agegrpname mapping was printed for both literacy and attendance_gender side by side:

literacy: agegrp=1 -> 05 Year, agegrp=7 -> 11 Year
attendance_gender: agegrp=5 -> 5 Year, agegrp=7 ->7 Year
attainment: agegrp=1-> 05 Year, agegrp=7 -> 11 Year

Looking at these value , we can determine that agegrp code  in literacy and attainment is not mapped same as in attendance_gender

Literacy and attainment uses agrgrp=age-4 i.e code 1 -> age 5 and so forth
while attendance_gender use the value as is.

#### 1.3.1.2 agegrp value problem
looking at the output above , in literacy and attainment,one code represent only one age only upto code 11, code 12 represent grouped age instead.
but in attendence_gender there is no grouped age ,and age code only goes upto 25

### 1.3.2 Fixing the 'agegrp'

To fix agegrp so its same in all the csv, just add agegrp with 4 to match the actual age.
make sure the initial age is geater than 1 and less than or equal to 11. 

```python
# Keep only single-year age codes (1-11 = ages 5-15), then shift by +4 to match attendance's coding

literacy_fixed = literacy[(literacy['agegrp'] >= 1) & (literacy['agegrp'] <= 11)].copy()
literacy_fixed['agegrp'] = literacy_fixed['agegrp'] + 4

attainment_fixed = attainment[(attainment['agegrp'] >= 1) & (attainment['agegrp'] <= 11)].copy()
attainment_fixed['agegrp'] = attainment_fixed['agegrp'] + 4

print(literacy_fixed[['agegrp','agegrpname']].drop_duplicates().sort_values('agegrp'))
print(attainment_fixed[['agegrp','agegrpname']].drop_duplicates().sort_values('agegrp'))
```

```python
# check again
merge_check = pd.merge(literacy_fixed, attendance_gender, on=['prov','dist','gapa','sex','agegrp'])
print("Merged shape:", merge_check.shape)

merge_check['attending_rate'] = merge_check['attnding'] / merge_check['rowtotal']
print(merge_check['attending_rate'].describe())
```

# 2. Merge the datframe

Merge all three dataframe on common keys ['prov','dist','gapa','sex','agegrp']
and drop unnecessary columns

```python
# updated merge
df_merged = pd.merge(
    literacy_fixed.rename(columns={'rowtotal': 'rowtotal_17'}),
    attainment_fixed.drop(columns=['provname','dname','gapaname','sexname','agegrpname']).rename(columns={'rowtotal': 'rowtotal_18'}),
    on=['prov','dist','gapa','sex','agegrp']
)

df_final = pd.merge(
    df_merged,
    attendance_gender,
    on=['prov','dist','gapa','sex','agegrp']
)

print("Final merged shape:", df_final.shape)
```

```python
df_final['attending_rate_check'] = df_final['attnding'] / df_final['rowtotal_20']
print(df_final['attending_rate_check'].describe())
```

```python
print(df_final.isnull().sum())
```

# 3. Remove aggegrate and null values

original census data constist of aggegrated data with code 0 for aggegrate , we dont need those 

```python
df_clean = df_final[(df_final['gapa'] != 0) & (df_final['sex'] != 0) & (df_final['agegrp'] != 0)].copy()
print("After filtering aggregates:", df_clean.shape)

df_clean = df_clean[(df_clean['rowtotal_17']!=0) & (df_clean['rowtotal_18']!=0) & (df_clean['rowtotal_20']!=0)].copy()
print("After filtering totals:", df_clean.shape)

```

# 4. Feature Engineering
new columns
illiteracy_rate=c_noreadwrite/rowtotal_17
high_illiteracy_risk= 1 if illiteracy_rate> median of illiteracy rate


```python
df_clean['illiteracy_rate']=df_clean['c_noreadwrite']/df_clean['rowtotal_17']
median_illiteracy_rate= df_clean['illiteracy_rate'].median()
df_clean['high_illiteracy_risk']= np.where(df_clean['illiteracy_rate']>median_illiteracy_rate,1,0)
```

```python
print(df_clean[['illiteracy_rate','high_illiteracy_risk']].describe())
print(df_clean['high_illiteracy_risk'].value_counts())
```

### more feature engineering
turning counts columns such as [a_earlyedu', 'b_primary', 'c_lowrsecndry', 'd_upprsecndry', 'e_slc', '] from table 18 into rate/ percentage by dividing with rowtotal_18
 note other are not included because no 15 year old wwould have graduated , slc count was included to addreess the minority , but no above level was addressed


```python
print(df_clean['e_slc'].describe())
print((df_clean['e_slc'] > 0).sum())

print(df_clean['f_inter'].describe())
print((df_clean['f_inter'] > 0).sum())
```

```python
# convert to rate
df_clean['early_edu_rate'] = df_clean['a_earlyedu'] / df_clean['rowtotal_18']
df_clean['primary_rate'] = df_clean['b_primary'] / df_clean['rowtotal_18']
df_clean['secondary_rate'] = (df_clean['c_lowrsecndry'] + df_clean['d_upprsecndry']) / df_clean['rowtotal_18']
df_clean['slc_rate'] = df_clean['e_slc'] / df_clean['rowtotal_18']
df_clean['no_education_rate'] = df_clean['j_nolevel'] / df_clean['rowtotal_18']

print(df_clean[['early_edu_rate','primary_rate','secondary_rate','slc_rate','no_education_rate']].describe())
```

## even more feature engineering

this is of table 20
turning count values like attnding, evrattnding, nvrattnding
to rate

```python
df_clean['attending_rate']=df_clean['attnding']/df_clean['rowtotal_20']
df_clean['ever_attending_rate']=df_clean['evrattnding']/df_clean['rowtotal_20']
df_clean['never_attending_rate']=df_clean['nvrattnding']/df_clean['rowtotal_20']
```

```python
print(df_clean[['attending_rate','ever_attending_rate','never_attending_rate']].describe())
```

# 1. Pre-Processing

## 1.1 Import necessary Library

```python
import pandas as pd
import numpy as np
```

## 1.2 Load CSV

```python
literacy = pd.read_csv('../data/table17_literacy.csv')
attainment = pd.read_csv('../data/table18_attainment.csv')
attendance= pd.read_csv('../data/table20_attendance.csv')

print("Table 17 shape:", literacy.shape)
print("Table 18 shape:", attainment.shape)
print("Table 20 shape:", attendance.shape)
```

## 1.2 Inspect the Data

```python
# Print out all column names of all three csv
print(literacy.columns.tolist())
print(attainment.columns.tolist())
print(attendance.columns.tolist())
```

### 1.2.1 Identifying merging columns

literacy and attainment both have column 'sex' with value male=1 and female=2, but attendence have seperate  column for gender(eg. male, female, attnding_m, attnding_f).

Merging all df on common keys ['prov','dist','gapa','sex','agegrp'] will fail because attendance has no 'sex' column to merge on

Fix:
Reshape attendance from wide format to long format.
so instead of multiple columns for gender, we have only one column.

### 1.2.2 Reshaping Attendance from wide to long format

```python
# reshaping 
# combine all male into a new df
attendance_male = attendance[['prov', 'dist', 'gapa', 'agegrp', 'male', 'attnding_m', 'evrattnding_m', 'nvrattnding_m', 'notstd_m']]
print(attendance_male.head())
attendance_female = attendance[['prov', 'dist', 'gapa', 'agegrp', 'female', 'attnding_f', 'evrattnding_f', 'nvrattnding_f', 'notstd_f']]
print(attendance_female.head())
```

### 1.2.3 Adding a New column named sex

```python
# male and female became rowtotal_20 to keep the total value, a new column is added to preserve the gender data
attendance_male.columns = ['prov', 'dist', 'gapa', 'agegrp', 'rowtotal_20', 'attnding', 'evrattnding', 'nvrattnding', 'notstd_20']
attendance_male['sex'] = 1

attendance_female.columns = ['prov', 'dist', 'gapa', 'agegrp', 'rowtotal_20', 'attnding', 'evrattnding', 'nvrattnding', 'notstd_20']
attendance_female['sex'] = 2
```

```python
 # check
print(attendance_male.head())
print(attendance_female.head())
```

### 1.2.4 Stack male and female datafrmae as one


```python

attendance_gender=pd.concat([attendance_male,attendance_female],ignore_index=True)
print("Shape of attendance gender", attendance_gender.shape)
attendance_gender
```

## 1.3 Another complication

1. Merging literacy and attendance_gender on common keys
2. Calculating attendance rate(attending/rowtotal), in theory it shoulndnt exceed 1.
3. making a new df called invalid rows which consists rows whose rate is > 1 to identify problem

```python
# Test merge literacy and attendance_gender on common keys
# merging literacy and attendance_gender on common keys
merge_test = pd.merge(literacy, attendance_gender, on=['prov','dist','gapa','sex','agegrp'])
print("Merged shape:", merge_test.shape)

#Calculate attendance rate that should be between 0 and 1
merge_test['attending_rate'] = merge_test['attnding'] / merge_test['rowtotal']
print(merge_test['attending_rate'].describe())

# Isolate rows where the rate is impossible (above 1)
invalid_rows = merge_test[merge_test['attending_rate'] > 1]
print("Number of rows with rate > 1:", invalid_rows.shape[0])
invalid_rows[['prov','dist','gapa','sex','agegrp','agegrpname','rowtotal','attnding']].head(10)
```

```python
print("Literacy agegrp mapping:")
print(literacy[['agegrp','agegrpname']].drop_duplicates().sort_values('agegrp'))

print("\nAttendance (original) agegrp mapping:")
print(attendance[['agegrp','agegrpname']].drop_duplicates().sort_values('agegrp'))
```
